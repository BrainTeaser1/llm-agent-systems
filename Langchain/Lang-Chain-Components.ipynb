```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
from langchain_openai import AzureChatOpenAI
import os

model = AzureChatOpenAI(
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY"),
    deployment_name="gpt-4o",
    api_version="2024-12-01-preview",
    temperature=0.9
)
```

```python
messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="What is are the benifits of using Langchain? Compare it with other frameworks.")
]
```

```python
response = model.invoke(messages)
print(response.content)
```

```python
for chunk in model.stream(messages):
    if chunk.content:
        print(chunk.content, end="", flush=True)
```

```python
from langchain_core.prompts import ChatPromptTemplate
translation_template = ChatPromptTemplate.from_messages([
    ("system", "You are Proffesional Translator. Translate the following {text} from {source_language} to {target_language}. Please Maintain the tone and style of the original text."),
    ("user", "{text}")
])
```

```python
prompt=translation_template.invoke({
    "source_language":"English",
    "target_language":"Hindi",
    "text":"Craving the decency of merciful lies."
})

response = model.invoke(prompt)
print(response.content)
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

def create_story_chain():
    story_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a creative story writer. Write a short and engaging story based on a given theme."),
        ("user", "Theme: {theme}\n Main character: {main_character}\n Setting: {setting}")
    ])

    analysis_prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a literary critic. Analyze the following story and provide insights on its themes, characters, and writing style."),
        ("user", "Story: {story}")
    ])

    story_chain = story_prompt | model | StrOutputParser()

    def add_story(input_dict):
        story = story_chain.invoke(input_dict)
        return {
            "story": story
        }

    def combine_output(input_dict):
        analysis = (analysis_prompt | model | StrOutputParser()).invoke(input_dict)
        return {
            "story": input_dict["story"],
            "analysis": analysis
        }

    chain = (
        RunnableLambda(add_story)
        | RunnableLambda(combine_output)
    )

    return chain
```

```python
chain=create_story_chain()
```

```python
result = chain.invoke({
    "theme": "Peaky Blinders-style crime drama set in 1920s Birmingham.Keep it a dark, emotional, morally complex drama (avoid explicit violence)",
    "main_character": "A Guy named Krishna Who is a law abbiding man but gets involved in crime due to circumstances.",
    "setting": "2020's , Delhi, India" 
})

print(result)
   
```

```python
print("\n===== STORY =====\n")
print(result["story"])

print("\n===== ANALYSIS =====\n")
print(result["analysis"])
```
